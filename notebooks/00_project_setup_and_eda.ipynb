# Bank Customer Churn Prediction Using Machine Learning

## Project Introduction
This project aims to predict whether a bank customer will leave the bank or not using supervised machine learning algorithms.

## Business Problem
Customer churn is an important business problem in banking because losing customers can reduce revenue and increase customer acquisition costs. By predicting churn in advance, banks can take action to retain customers.

## Dataset Source
Dataset: Bank Customer Churn Prediction  
Source: Kaggle  
Link: https://www.kaggle.com/datasets/shantanudhakadd/bank-customer-churn-prediction

## Objective
The objective of this notebook is to:
- load and inspect the dataset
- understand its structure
- check for missing values and duplicates
- analyze the target variable
- perform exploratory data analysis (EDA)

Cell 2 — Import libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# display settings
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 1000)

# plotting style
sns.set_style("whitegrid")
plt.rcParams["figure.figsize"] = (8, 5)
```

```python
BASE_DIR = Path("..").resolve()
DATA_PATH = BASE_DIR / "data" / "raw" / "Churn_Modelling.csv"
FIGURES_DIR = BASE_DIR / "outputs" / "figures"

FIGURES_DIR.mkdir(parents=True, exist_ok=True)

DATA_PATH
```

## EDA Summary

Key observations from the dataset:

- The dataset contains 10,000 customer records and 14 columns.
- The target variable is `Exited`, where:
  - `0` = customer stayed
  - `1` = customer left the bank
- There are no missing values in the dataset.
- There are no exact duplicate rows.
- The target classes are imbalanced, with non-churn customers forming the majority.
- `Geography`, `Gender`, and possibly `Age` appear to show meaningful variation with churn.
- `RowNumber`, `CustomerId`, and `Surname` are not useful predictive features and should be removed during preprocessing.
